```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load main dataset
df = pd.read_csv('../data/learn_dataset.csv')
df.head()
```

## 1. Missing Values & Basic Info
- Check for missing values, data types, and basic statistics.

```python
df.info()
df.isnull().sum()
```

## 2. Feature Analysis
- Correlation matrix, feature importance (mutual info), frequency analysis.

```python
from sklearn.feature_selection import mutual_info_classif
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True)
plt.show()

# Mutual information
X = df.drop(['Target'], axis=1, errors='ignore')
y = df['Target'] if 'Target' in df.columns else None
if y is not None:
    mi = mutual_info_classif(X.select_dtypes(include=[np.number]).fillna(0), y)
    mi_series = pd.Series(mi, index=X.select_dtypes(include=[np.number]).columns)
    mi_series.sort_values(ascending=False).plot(kind='bar')
    plt.title('Mutual Information (Numerical Features)')
    plt.show()
```

## 3. Preprocessing
- Encode categoricals, normalize, handle outliers.

```python
from sklearn.preprocessing import OrdinalEncoder, StandardScaler
cat_cols = df.select_dtypes(include=['object']).columns
df[cat_cols] = OrdinalEncoder().fit_transform(df[cat_cols])

if 'is_student' in df.columns:
    df['is_student'] = df['is_student'].astype(int)

scaler = StandardScaler()
num_cols = df.select_dtypes(include=[np.number]).columns
df[num_cols] = scaler.fit_transform(df[num_cols])
df.head()
```
